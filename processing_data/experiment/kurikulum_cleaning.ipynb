```python
import pandas as pd
from pathlib import Path
import json
```

```python
main_folder = Path.cwd().parent

df = pd.read_excel(f"{main_folder}/data/excels/kurikulum-revisi-24-cleaned.xlsx")
```

```python
df.head()
```

```python
df = df.iloc[:, :6]

df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df = df.dropna()
```

```python
df['SKS'] = df['SKS'].astype(int)
df['semester'] = df['semester'].astype(int)

df.info()
```

```python
def data_kosong(kolom):
    return f"Data {kolom} belum tersedia"

dokumen = []

for _, row in df.iterrows():
    kode_matkul = row["kode_mata_kuliah"] if row["kode_mata_kuliah"] != "NONE" else data_kosong("kode mata kuliah")
    nama_matkul = row["nama_mata_kuliah"] if row["nama_mata_kuliah"] != "NONE" else data_kosong("nama mata kuliah")
    SKS = row["SKS"] if row["SKS"] != "NONE" else data_kosong("SKS")
    semester = row["semester"] if row["semester"] != "NONE" else data_kosong("semester")
    status = row["status"] if row["status"] != "NONE" else data_kosong("status")
    deskripsi = row["deskripsi"] if row["deskripsi"] != "NONE" else data_kosong("deskripsi")
    deskripsi_tersedia = True if row['deskripsi'] != "NONE" else False

    baris = f"Mata kuliah: {nama_matkul} \n Kode mata kuliah: {kode_matkul} \n Semester: {semester} \n Bobot: {SKS} SKS \n Jenis mata kuliah: {status} \n Deskripsi: {deskripsi}"
    teks = f"{nama_matkul} ({kode_matkul}) adalah mata kuliah semester {semester} di Program Studi Teknik Komputer dengan bobot {SKS} SKS. \n {baris}"

    meta = {
        "tipe": "detail_matkul_24_rev",
        "semester": semester,
        "kode_matkul": kode_matkul,
        "nama_matkul": nama_matkul,
        "sks": SKS,
        "jenis_matkul": status,
        "deskripsi_tersedia": deskripsi_tersedia
    }

    data_jadi = {
        "id" : f"matkul-{kode_matkul}",
        "text" : teks,
        "metadata" : meta
    }

    dokumen.append(data_jadi)

dokumen[:5]
    
        
       
```

```python
with open(f"{main_folder}\data\jsonl\kurikulum-24-rev.jsonl", 'w', encoding='utf-8') as f:
    for d in dokumen:
        f.write(json.dumps(d, ensure_ascii=False) + "\n")
```
