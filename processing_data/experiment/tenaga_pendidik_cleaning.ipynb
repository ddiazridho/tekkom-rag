```python
main_folder = Path.cwd().parent

df = pd.read_excel(f"{main_folder}/data/excels/tenaga-pendidik-cleaned.xlsx")

df.head()
```

```python
standard_column = ['nama', 'jabatan']

df.columns = standard_column 

df.head()
```

```python
gelar = ["Prof", "Dr", "Drs", "Dra", "Ir"]

def hapus_gelar(nama):
    nama = nama.split(',')[0]
    nama = nama.split(' ')
    nama = [item for item in nama if item not in gelar]
    return ' '.join(nama)
```

```python
def ubah_format_nama(nama):
    return nama.title()
```

```python
dokumen = []
prodi = 'Teknik Komputer'

for idx, row in df.iterrows():
    nama = ubah_format_nama(row['nama'])
    nama_tanpa_gelar = hapus_gelar(ubah_format_nama(row['nama']))
    jabatan = row['jabatan']

    baris = f"Nama : {nama} \n Nama Tanpa Gelar : {nama_tanpa_gelar} \n Jabatan : {jabatan} "
    teks = f"{nama} adalah seorang tenaga pendidik yang memegang jabatan sebagai {jabatan}. \n\n {baris}"

    metadata = {
        "tipe" : "tenaga pendidik",
        "nama" : nama,
        "jabatan" : jabatan
    }

    dokumen.append({
        'id' : f"{idx}-tendik-{slug(nama_tanpa_gelar)}",
        'text' : teks,
        'metadata' : metadata
    })    
```

```python
with open(f"{main_folder}\data\jsonl\\tendik-tekkom.jsonl", 'w', encoding='utf-8') as f:
    for d in dokumen:
        f.write(json.dumps(d, ensure_ascii=False) + "\n")
```
