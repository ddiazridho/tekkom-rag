```python
import os
import json
import re
from pathlib import Path

import pandas as pd
```

### **2. DATA LOADING**

```python
main_folder = Path.cwd().parent
path_excel = f"{main_folder}/data/excels/dosen-dalam-tekkom-cleaned.xlsx"

df = pd.read_excel(path_excel)

df.head()
```

```python
df.info()
```

```python
gelar = {'Prof.', 'Dr.', 'Ir.', 'S.T.', 'M.T.', 'M.Sc.', 'M.Eng.', 'M.Kom.', 'M.Kom', 'S.Kom', 'S.Kom.', 'S.Pd', 'S.Pd.', 'S.Si', 'S.Si.', 'S.E', 'S.E.', 'S.H', 'S.H.', 'S.Ars', 'S.Ars.', 'M.Ars', 'M.Ars.', 'M.Sn', 'M.Sn.'}

def nama_tanpa_gelar(nama):
    nama = nama.split(',')[0].split(' ')

    nama = [kata for kata in nama if kata not in gelar]
    nama = ' '.join(nama)

    return nama

nama_tanpa_gelar('Prof. Dr. Ir. John Doe, S.T., M.T.')
```

```python
def slug(teks):
    teks = teks.split(' ')
    return '-'.join(teks).lower()
```

```python
kategori = 'dalam'
NAMA_PRODI = "Teknik Komputer"
sumber_file = "dosen-dalam-tekkom-cleaned.xlsx"

dokumen = []

for index, row in df.iterrows():
    nama_dosen_tanpa_gelar = nama_tanpa_gelar(row['nama_dosen_dalam_tekkom'])
    keahlian = row['keahlian']

    baris = [f"Nama: {row['nama_dosen_dalam_tekkom']}", f"Nama tanpa gelar: {nama_dosen_tanpa_gelar}", f"Kategori: Dosen {kategori}"]
    teks = (f"{row['nama_dosen_dalam_tekkom']} adalah dosen {kategori} di Program Studi {NAMA_PRODI}.\n\n"
                + "\n".join(baris))

    meta = {"tipe": "profil_dosen", "kategori_dosen": kategori, "nama": row["nama_dosen_dalam_tekkom"],
                "sumber_file": sumber_file}
    profiles = ['foto_profil', 'profil_sinta', 'publikasi', 'profil_undip']

    for profile in profiles:
        meta[profile] = row[profile]

    dokumen.append({"teks": teks, "meta": meta, "slug": f"dosen-{kategori}-{slug(nama_dosen_tanpa_gelar)}"})
```

```python
dokumen[:3]
```

```python
with open(f"{main_folder}\data\jsonl\dosen_dalam_tekkom.jsonl", 'w', encoding='utf-8') as f:
    for d in dokumen:
        f.write(json.dumps(d, ensure_ascii=False) + "\n")
```
